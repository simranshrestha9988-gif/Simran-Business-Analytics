  <h2 style="color: #006699; font-weight: bold;">SELECT</h2>

  <h3 style="color: #006699;">Definition</h3>
  <p><strong>SELECT</strong> retrieves data from a table. Name the columns to retrieve, or use <code>*</code> to retrieve all columns.</p>

  <h3 style="color: #006699;">Syntax</h3>
  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">
SELECT column1, column2 FROM table_name;
SELECT * FROM table_name;        -- * means all columns</pre>

  <h3 style="color: #006699;">Task: Read a CSV File</h3>
  <p>Read <strong>sales.csv</strong> into a pandas DataFrame and save it as a database table named <strong>sales</strong>.</p>

  <h3 style="color: #006699;">What This Task Does</h3>
  <ul>
    <li>Reads the CSV file using <code>pd.read_csv()</code>.</li>
    <li>Saves the data to the database using <code>to_sql()</code>.</li>
    <li>Retrieves all columns or selected columns using <code>SELECT</code>.</li>
  </ul>

  <h3 style="color: #006699;">Examples</h3>
  <ul>
    <li><strong>Example 1: All columns:</strong> <code>SELECT * FROM sales;</code></li>
    <li><strong>Example 2: Selected columns:</strong> <code>SELECT product, region, quantity FROM sales;</code></li>
  </ul>

</div>

```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load sales.csv into the 'sales' table
pd.read_csv('sales.csv').to_sql('sales', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# SELECT * returns every column.
# Example 1: All columns
pd.read_sql_query('''
SELECT * FROM sales
''', conn)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# Example 2: Only the columns you need
pd.read_sql_query('''
SELECT product, region, quantity FROM sales
''', conn)
```

  <h2 style="color: #006699; font-weight: bold;">WHERE</h2>

  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">
SELECT columns FROM table_name WHERE condition;

-- Comparison: =   &gt;   &lt;   &gt;=   &lt;=   &lt;&gt;  (not equal)
-- Combine: AND   OR
-- Text values go in quotes: 'Sales'
-- Numbers do not: 60000</pre>

  <h2 style="color: #006699; font-weight: bold;">GROUP BY</h2>

  <h3 style="color: #006699;">Definition</h3>
  <p><strong>GROUP BY</strong> groups rows with matching values so that each group can be summarized. Similar to an Excel PivotTable, it is commonly used with aggregate functions such as <code>SUM</code> or <code>COUNT</code>.</p>

  <h3 style="color: #006699;">Syntax</h3>
  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">
SELECT group_column, SUM(value_column) AS total
FROM table_name
GROUP BY group_column;</pre>

```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load marketing_campaigns.csv into the 'campaigns' table
pd.read_csv('marketing_campaigns.csv').to_sql('campaigns', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# GROUP BY combines records into groups for summary calculations.
# SUM() adds values.
# AS gives a result column or table a short name.
# Example 1: Total spend by channel
pd.read_sql_query('''
SELECT channel, SUM(spend) AS total_spend
FROM campaigns
GROUP BY channel
''', conn)
```

  <h2 style="color: #006699; font-weight: bold;">HAVING</h2>

  <h3 style="color: #006699;">Definition</h3>
  <p><strong>HAVING</strong> filters groups after <code>GROUP BY</code>. Use <code>WHERE</code> for conditions on individual rows and <code>HAVING</code> for conditions on totals, counts, or averages.</p>

  <h3 style="color: #006699;">Syntax</h3>
  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">
SELECT group_column, SUM(value_column) AS total
FROM table_name
GROUP BY group_column
HAVING SUM(value_column) &gt; 100;</pre>

  <h3 style="color: #006699;">Task: Read a CSV File</h3>
  <p>Read <strong>restaurant_orders.csv</strong> into a pandas DataFrame and save it as a database table named <strong>orders</strong>.</p>

  <h3 style="color: #006699;">What This Task Does</h3>
  <ul>
    <li>Reads restaurant orders using <code>pd.read_csv()</code>.</li>
    <li>Saves the data to the database using <code>to_sql()</code>.</li>
    <li>Groups records by server or menu item using <code>GROUP BY</code>.</li>
    <li>Filters groups by total tips or number of orders using <code>HAVING</code>.</li>
  </ul>

  <h3 style="color: #006699;">Examples</h3>
  <ul>
    <li><strong>Example 1: Servers with more than $130 in tips:</strong> <code>SELECT server, SUM(tip) AS total_tips FROM orders GROUP BY server HAVING SUM(tip) &gt; 130;</code></li>
    <li><strong>Example 2: Menu items ordered at least 6 times:</strong> <code>SELECT menu_item, COUNT(*) AS times_ordered FROM orders GROUP BY menu_item HAVING COUNT(*) &gt;= 6;</code></li>

```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load restaurant_orders.csv into the 'orders' table
pd.read_csv('restaurant_orders.csv').to_sql('orders', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# GROUP BY combines records into groups for summary calculations.
# HAVING keeps groups that meet the aggregate condition.
# SUM() adds values.
# ROUND(value, digits) rounds a number to the specified decimal places.
# AS gives a result column or table a short name.
# Example 1: Servers with more than $130 in tips
pd.read_sql_query('''
SELECT server, ROUND(SUM(tip), 2) AS total_tips
FROM orders
GROUP BY server
HAVING SUM(tip) > 130
''', conn)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# GROUP BY combines records into groups for summary calculations.
# HAVING keeps groups that meet the aggregate condition.
# COUNT(*) counts rows.
# AS gives a result column or table a short name.
# Example 2: Menu items ordered at least 6 times
pd.read_sql_query('''
SELECT menu_item, COUNT(*) AS times_ordered
FROM orders
GROUP BY menu_item
HAVING COUNT(*) >= 6
''', conn)
```

  <h2 style="color: #006699; font-weight: bold;">
    Aggregate Functions
  </h2>

  <h3 style="color: #006699;">Definition</h3>
  <p>
    <strong>Aggregate functions</strong> calculate one value
    from many rows:
    <code>SUM()</code> adds,
    <code>COUNT()</code> counts rows,
    <code>AVG()</code> calculates the average,
    <code>MIN()</code> finds the smallest value, and
    <code>MAX()</code> finds the largest value.
    Use <code>ROUND(value, 2)</code> to round decimals.
  </p>

  <h3 style="color: #006699;">Syntax</h3>

  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">SELECT COUNT(*), SUM(col), AVG(col), MIN(col), MAX(col)
FROM table_name;</pre>

  <p>
    <strong>Note:</strong> GROUP BY produces one summary
    for each group.
  </p>

    <code>conn</code>.
  </p>

  <h3 style="color: #006699;">What This Task Does</h3>

  <ul>
    <li>Loads clinic visits into a database table.</li>
    <li>Calculates counts, averages, minimums, and maximums.</li>
    <li>Summarizes clinic performance by department.</li>
  </ul>

  <h3 style="color: #006699;">Examples</h3>

  <p><strong>Example 1: Whole Clinic Summary</strong></p>

  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">SELECT COUNT(*) AS total_visits,
       AVG(wait_minutes) AS avg_wait,
       MIN(visit_cost) AS lowest_cost,
       MAX(visit_cost) AS highest_cost
FROM visits;</pre>

  <p><strong>Example 2: Summary by Department</strong></p>

  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">SELECT department,
       COUNT(*) AS visits,
       ROUND(AVG(visit_cost), 2) AS avg_cost
FROM visits
GROUP BY department;</pre>

</div>


```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load clinic_visits.csv into the 'visits' table
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# COUNT(*) counts rows.
# AVG() calculates the average of nonmissing values.
# ROUND(value, digits) rounds a number to the specified decimal places.
# MIN() returns the smallest nonmissing value.
# MAX() returns the largest nonmissing value.
# AS gives a result column or table a short name.
# Example 1: Whole clinic summary
pd.read_sql_query('''
SELECT COUNT(*) AS total_visits,
       ROUND(AVG(wait_minutes), 1) AS avg_wait_minutes,
       MIN(visit_cost) AS lowest_cost,
       MAX(visit_cost) AS highest_cost
FROM visits
''', conn)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# GROUP BY combines records into groups for summary calculations.
# COUNT(*) counts rows.
# AVG() calculates the average of nonmissing values.
# ROUND(value, digits) rounds a number to the specified decimal places.
# AS gives a result column or table a short name.
# Example 2: Summary by department
pd.read_sql_query('''
SELECT department,
       COUNT(*) AS visits,
       ROUND(AVG(visit_cost), 2) AS avg_cost
FROM visits
GROUP BY department
''', conn)
```

  <h2 style="color: #006699; font-weight: bold;">JOIN</h2>

  <h3 style="color: #006699;">Definition</h3>
  <p>
    <strong>JOIN</strong> combines rows from two tables
    using a shared column.
    <code>INNER JOIN</code> keeps only matching rows.
    <code>LEFT JOIN</code> keeps all rows from the left table,
    even when no match exists.
  </p>

  <h3 style="color: #006699;">Syntax</h3>

  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">SELECT a.column, b.column
FROM table_a AS a
JOIN table_b AS b
ON a.key = b.key;</pre>

  <p>
    <strong>Note:</strong> LEFT JOIN keeps unmatched rows
    from the left table.
  </p>

    <strong>customers</strong> database table using
    <code>to_sql()</code> and the existing connection
    <code>conn</code>.
  </p>

  <p>
    Read <strong>bank_transactions.csv</strong> using
    <code>pd.read_csv()</code> and save the data as the
    <strong>transactions</strong> database table using
    <code>to_sql()</code> and <code>conn</code>.
  </p>

  <h3 style="color: #006699;">What This Task Does</h3>

  <ul>
    <li>Loads customers and transactions into separate database tables.</li>
    <li>Matches transactions with customer information using <code>customer_id</code>.</li>
    <li>Identifies transactions without matching customers.</li>
  </ul>

  <h3 style="color: #006699;">Examples</h3>

  <p><strong>Example 1: Transactions with Customer Names</strong></p>

  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">SELECT c.customer_name,
       c.branch,
       t.transaction_type,
       t.amount
FROM transactions AS t
JOIN customers AS c
ON t.customer_id = c.customer_id;</pre>

  <p><strong>Example 2: Transactions Without Matching Customers</strong></p>

  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">SELECT t.transaction_id,
       t.customer_id,
       c.customer_name,
       t.amount
FROM transactions AS t
LEFT JOIN customers AS c
ON t.customer_id = c.customer_id
WHERE c.customer_id IS NULL;</pre>

</div>


```python
# read_csv() reads a CSV file into a pandas table.
# to_sql() copies the table into the connected database.
# if_exists="replace" replaces the existing table with the CSV data.
# index=False avoids saving the pandas row index as a column.
# Load both files into their own tables
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists='replace', index=False)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# AS gives a result column or table a short name.
# JOIN matches rows from tables using the ON condition.
# Example 1: Transactions with customer names
pd.read_sql_query('''
SELECT c.customer_name, c.branch, t.transaction_type, t.amount
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
''', conn)
```

```python
# read_sql_query() runs SQL and returns the results as a pandas DataFrame.
# Triple quotation marks allow SQL to span several lines.
# conn identifies the database connection.
# WHERE keeps rows that meet the condition.
# AS gives a result column or table a short name.
# JOIN matches rows from tables using the ON condition.
# LEFT JOIN also keeps rows from the left table without a match.
# IS NULL checks for a missing value.
# Example 2: Transactions with no matching customer (LEFT JOIN)
pd.read_sql_query('''
SELECT t.transaction_id, t.customer_id, c.customer_name, t.amount
FROM transactions AS t
LEFT JOIN customers AS c ON t.customer_id = c.customer_id
WHERE c.customer_name IS NULL
''', conn)
```

  <h2 style="color: #006699; font-weight: bold;">Key Takeaways</h2>
  <ul>
    <li><strong>SELECT</strong> asks for columns; <strong>WHERE</strong> filters rows; <strong>ORDER BY</strong> sorts.</li>
    <li><strong>GROUP BY</strong> + an aggregate is a pivot table in text form; <strong>HAVING</strong> filters the groups.</li>
    <li><strong>JOIN</strong> connects tables through a shared key: use <strong>LEFT JOIN</strong> when you must not lose rows.</li>
    <li>Loading a CSV with <code>to_sql()</code> and querying with <code>read_sql_query()</code> is the same pattern for every dataset.</li>
  </ul>

  <h2 style="color: #006699;">Query order to remember</h2>
  <pre style="background: #f4f4f4; padding: 10px; border-left: 4px solid #006699;">
SELECT ... FROM ... [JOIN ... ON ...] WHERE ... GROUP BY ... HAVING ... ORDER BY ... LIMIT ...</pre>

```python
print(pd.read_csv('bank_customers.csv').columns)
print(pd.read_csv('bank_transactions.csv').columns)
```

```python
pd.read_sql_query("""
SELECT bank_customers.branch,
       SUM(bank_transactions.amount) AS total_amount
FROM bank_customers
JOIN bank_transactions
ON bank_customers.customer_id = bank_transactions.customer_id
GROUP BY bank_customers.branch
""", conn)
```
