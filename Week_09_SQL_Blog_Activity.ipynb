# **Task 1: SELECT - Sales Data

```python
pd.read_csv('sales.csv').to_sql('sales', conn, if_exists='replace', index=False)
pd.read_csv('hr_employees.csv').to_sql('hr_employees', conn, if_exists='replace', index=False)
pd.read_csv('realestate_listings.csv').to_sql('realestate_listings', conn, if_exists='replace', index=False)
pd.read_csv('marketing_campaigns.csv').to_sql('marketing_campaigns', conn, if_exists='replace', index=False)
pd.read_csv('restaurant_orders.csv').to_sql('restaurant_orders', conn, if_exists='replace', index=False)
pd.read_csv('clinic_visits.csv').to_sql('clinic_visits', conn, if_exists='replace', index=False)
pd.read_csv('bank_customers.csv').to_sql('bank_customers', conn, if_exists='replace', index=False)
pd.read_csv('bank_transactions.csv').to_sql('bank_transactions', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query("""
SELECT sale_date, product, price
FROM sales
""", conn)
```

# **Task 2: WHERE - Human Resources

```python
pd.read_sql_query("""
SELECT *
FROM hr_employees
WHERE city = 'Chicago' AND years_experience > 10
""", conn)
```

# **Task 4: GROUP BY — Marketing

```python
pd.read_csv('marketing_campaigns.csv').columns
```

```python
pd.read_sql_query("""
SELECT week, SUM(impressions) AS total_impressions
FROM marketing_campaigns
GROUP BY week
""", conn)
```

# **Task 5: HAVING — Restaurant

```python
pd.read_csv('restaurant_orders.csv').columns
```

```python
pd.read_sql_query("""
SELECT order_date, SUM(order_total) AS total_sales
FROM restaurant_orders
GROUP BY order_date
HAVING SUM(order_total) > 400
""", conn)
```

# **Task 6: Aggregate Functions — Healthcare

```python
pd.read_csv('clinic_visits.csv').columns
```

```python
pd.read_sql_query("""
SELECT doctor,
       AVG(patient_age) AS average_age,
       MAX(wait_minutes) AS longest_wait
FROM clinic_visits
GROUP BY doctor
""", conn)
```

# **Task 7: JOIN — Banking

```python
print(pd.read_csv('bank_customers.csv').columns)
print(pd.read_csv('bank_transactions.csv').columns)
```

```python
pd.read_sql_query("""
SELECT bank_customers.branch,
       SUM(bank_transactions.amount) AS total_amount
FROM bank_customers
JOIN bank_transactions
ON bank_customers.customer_id = bank_transactions.customer_id
GROUP BY bank_customers.branch
""", conn)
```
